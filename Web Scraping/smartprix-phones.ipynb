```python
import numpy as np 
import pandas as pd 
from bs4 import BeautifulSoup
```

```python
with open('htmls/smartprix.html', 'r', encoding='utf-8') as f: 
    html = f.read()
```

```python
soup = BeautifulSoup(html, 'lxml')
print(soup)
```

```python
phone_cards = soup.find_all('div', class_='sm-product has-tag has-features has-actions')
```

```python
len(phone_cards)
```

```python
count = 1
for phone in phone_cards: 
    try: 
        print(f'{count}. {phone.find('h2').text}')
    except: 
         print(f'{count}. {np.nan}')
    count +=1
```
